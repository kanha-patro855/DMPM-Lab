```python
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression
```

```python
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv('/content/p_score.csv')
df
```

```python
plt.scatter(df['IQ'],df['Placement Score'])
plt.xlabel('IQ')
plt.ylabel('Placement Score')
plt.show()
```

```python
plt.scatter(df['CGPA'],df['Placement Score'])
plt.xlabel('CGPA')
plt.ylabel('Placement Score')
plt.show()
```

```python
reg=LinearRegression()
reg.fit(df[['IQ','CGPA']],df['Placement Score'])
reg
```
