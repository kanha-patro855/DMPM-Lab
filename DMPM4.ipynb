```python
# 1.Implement the min max normalisation on a set of datapoints
points=[8,20,12,10]
normalised=[]
min=min(points)
max=max(points)
for i in points:
    norm=(i-min)/(max-min)
    normalised.append(norm)
print(normalised)
```

```python
# 2.Implement the min max normalisation on a dataset
import pandas as pd
import numpy as np
df=pd.read_csv("/content/winequality-red.csv")
normalised=[]
df.rename(columns={'fixed acidity':'fixed_acidity','volatile acidity':'volatile_acidity','citric acid':'citric_acid','residual sugar':'residual_sugar','free sulfur dioxide':'free_sulfur_dioxide','total sulfur dioxide':'total_sulfur_dioxide'},inplace=True)
for i in df.columns:
    min=df[i].min()
    max=df[i].max()
    for j in df[i]:
        norm=(j-min)/(max-min)
        normalised.append(norm)
normalised_array=np.array(normalised)
normalised_array
```

```python
points=[7,15,25,99,150]
normalised=[]

# Find the maximum absolute value in the dataset
max_val = max(abs(p) for p in points)

# Determine the smallest power of 10 greater than or equal to max_val
j = 0
while 10**j <= max_val:
    j += 1
max_divisor = 10**j

for i in points:
    norm = i / max_divisor
    normalised.append(norm)
print(normalised)
```

```python
# correlation co-efficient relation and co variance analysis.

x_data = [10,20,30,40,50]
y_data = [12,24,33,45,55]

def mean(data):
    return sum(data) / len(data)

def calculate_covariance(x, y):
    n = len(x)
    mean_x, mean_y = mean(x), mean(y)
    return sum((x[i] - mean_x) * (y[i] - mean_y) for i in range(n)) / (n - 1)

def calculate_std(data):
    n = len(data)
    mean_data = mean(data)
    variance = sum((val - mean_data) ** 2 for val in data) / (n - 1)
    return variance ** 0.5

def calculate_correlation(x, y):
    cov = calculate_covariance(x, y)
    std_x = calculate_std(x)
    std_y = calculate_std(y)
    return cov / (std_x * std_y)

cov = calculate_covariance(x_data, y_data)
corr = calculate_correlation(x_data, y_data)

print(f"Covariance: {cov:.2f}")
print(f"Correlation: {corr:.4f}")
```
