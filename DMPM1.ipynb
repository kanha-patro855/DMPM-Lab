```python
# 1Find sum and average

nums=list(map(int, input("enter numbers separated by space: ").split()))
sum_val=0

for num in nums:
    sum_val+=num

avg_val=sum_val/len(nums)
print("Sum is",sum_val,"and average is",avg_val)
```

```python
# 2Count no. of vowels and consonants

#text = input("Enter a string: ")
v_count,c_count=0,0

for ch in input('Enter a string: '):
    if ch in "aeiouAEIOU":
        v_count+=1
    else:
       c_count+=1

print(f"Number of vowels are: {v_count} and number of consonants are: {c_count}")
```

```python
# 3Find largest element in a list

l1=eval(input("Enter a list: "))
largest=l1[0]

for i in l1:
    if i>largest:
        largest=i

print("Largest element is:",largest)
```

```python
# 4Count even and odd

nums=[1,23,54,2,65,678]
even_count,odd_count=0,0

for num in nums:
    if num % 2==0:
        even_count+=1
    else:
        odd_count+=1

print(f"There are {even_count} evens and {odd_count} odds in the list.")
```

```python
# 5Reverse a string without using slicing

str=input("enter a string: ")
rev_str=""

for ch in str:
    rev_str=ch+rev_str

print(f"Reverse string is: {rev_str}")
```

```python
# 6Check prime

num=int(input("Enter a number: "))

if num>1:
    for i in range(2,num):
        if num%i==0:
            print("It is not a prime number.")
            break
    else:
        print("It is a prime number.")
else:
    print("It is not a prime number.")
```

```python
# 7Find factorial

def factorial(num):
  fact=1
  for i in range(1,num+1):
    fact*=i
  return fact

print(factorial(0))
factorial(5)
```

```python
# 8Sum of all elements in list

nums=eval(input("Enter a list: "))

for num in nums:
    sum_val+=num
print("Sum of all elements is:",sum_val)
```

```python
# 9Frequency of element in a list

l1=eval(input("Enter a list: "))
element=eval(input("Enter an element to count frequency: "))

count=0
for i in l1:
    if i==element:
        count+=1
print("Frequency of",element,"is:",count)
```

```python
# 10find maximum and minimum

l1=[1,23,453,12,45,2,2]
max=l1[0]
min=l1[0]

for i in l1:
    if i>max:
        max=i
    if i<min:
        min=i

print(f"Maximum is {max} and minimum is {min}")
```

```python
# 13PrintTable

def table(num):
  for i in range(1,11):
    print(f"{num}x{i} = {num*i}")

table(9)
```

```python
# 14Check plaindrome

def check_palindrome(str):
  if str==str[::-1]:
    print(f"\"{str}\" is palindrome.")
  else:
    print(f"\"{str}\" is not a palindrome.")

check_palindrome("qwertytrewq")
```
